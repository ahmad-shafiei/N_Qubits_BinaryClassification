# 4 Qubits

```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
from qiskit_aer.noise import NoiseModel, ReadoutError

import numpy as np
from itertools import product
```

```python
def build_circuit_4q(thetas):
    qc = QuantumCircuit(4, 4)
    
    for i in range(4):
        qc.ry(thetas[i], i)
    
    qc.measure([0,1,2,3], [0,1,2,3])
    return qc
```

```python
# Circuit
thetas = np.random.uniform(0, np.pi, 4)  # thetas =  [1.2, 2.2, 3, 1]
qc = build_circuit_4q(thetas)
# رسم مدار
qc.draw("mpl")
```

```python
def counts_to_probs_4q(counts):
    probs = np.zeros(16)
    
    for bitstring, count in counts.items():
        idx = int(bitstring, 2)
        probs[idx] = count
        
    probs = probs / np.sum(probs)
    return probs
```

```python
ideal_sim = AerSimulator()
noisy_sim = AerSimulator(noise_model=noise_model)


def get_ideal_4q(thetas, shots=4096):
    qc = build_circuit_4q(thetas)
    result = ideal_sim.run(qc, shots=shots).result()
    counts = result.get_counts()
    return counts_to_probs_4q(counts)

def get_noisy_4q(thetas, shots=4096):
    qc = build_circuit_4q(thetas)
    result = noisy_sim.run(qc, shots=shots).result()
    counts = result.get_counts()
    return counts_to_probs_4q(counts)

readout_error = ReadoutError([
    [0.90, 0.10],
    [0.15, 0.85]
])

noise_model = NoiseModel()

# اعمال روی هر 4 کیوبیت
for q in range(4):
    noise_model.add_readout_error(readout_error, [q])
```

```python
# example of applying stochastic matrix : A
# A4 = A ⊗  A ⊗ A ⊗  A
# p_noisy=A4.p_ideal
A = np.array([       ## this is exactly readout error but for 1 qubit
    [0.90, 0.10],
    [0.15, 0.85]
])

# for to qubits: 
A4 = np.kron(np.kron(np.kron(A, A), A), A)    # for 4 qubits
print("Shape A4:", A4.shape)   # (16,16)

# p_ideal = np.array([0.8, 0.1, 0.1, 0.0])
# random probability distribution
p_ideal = np.random.rand(16)
p_ideal = p_ideal / np.sum(p_ideal)   # normalize

p_noisy = A4 @ p_ideal

print('A: \n', A)

# print('A4: \n', A4)
print('p_noisy: \n', p_noisy)
# np.linalg.inv(A4)
```

### Generate dataset

```python
thetas = np.random.uniform(0, np.pi, 4)

p_i = get_ideal_4q(thetas)
p_n = get_noisy_4q(thetas)

print("sum ideal:", np.sum(p_i))
print("sum noisy:", np.sum(p_n))
```

```python
def generate_dataset_4q(num_samples=5000):
    
    X = []
    Y = []
    
    for _ in range(num_samples):
        thetas = np.random.uniform(0, np.pi, size=4)
        
        p_ideal = get_ideal_4q(thetas)
        p_noisy = get_noisy_4q(thetas)
        
        X.append(p_noisy)
        Y.append(p_ideal)
    
    return np.array(X), np.array(Y)
```

## MLP

```python
import torch
import torch.nn as nn
# from torch.optim import optimizer
import torch.optim as optim
from torch.utils.data import DataLoader, TensorDataset
import numpy as np
```

```python
class MLP_4Q(nn.Module):
    def __init__(self):
        super().__init__()
        
        self.net = nn.Sequential(
            nn.Linear(16, 128),
            nn.ReLU(),
            
            nn.Linear(128, 256),
            nn.ReLU(),
            
            nn.Linear(256, 128),
            nn.ReLU(),
            
            nn.Linear(128, 16),
            nn.Softmax(dim=1)
        )
    
    def forward(self, x):
        return self.net(x)
```

### Train

```python
X, Y = generate_dataset_4q(5000)

X_tensor = torch.tensor(X, dtype=torch.float32)
Y_tensor = torch.tensor(Y, dtype=torch.float32)

dataset = TensorDataset(X_tensor, Y_tensor)
train_loader = DataLoader(dataset, batch_size=64, shuffle=True)

model = MLP_4Q()

criterion = nn.KLDivLoss(reduction="batchmean")
optimizer = optim.Adam(model.parameters(), lr=1e-3)

epochs = 100
loss_history = []

for epoch in range(epochs):
    total_loss = 0
    
    for x_batch, y_batch in train_loader:
        
        optimizer.zero_grad()
        
        preds = model(x_batch)
        loss = criterion(torch.log(preds + 1e-8), y_batch)
        
        loss.backward()
        optimizer.step()
        
        total_loss += loss.item()
    
    avg_loss = total_loss / len(train_loader)
    loss_history.append(avg_loss)
    
    print(f"Epoch {epoch+1}, Loss: {avg_loss:.6f}")
```

```python
model.eval()

with torch.no_grad():
    x_test = X_tensor[0].unsqueeze(0)
    y_pred = model(x_test)

print("Noisy :", X_tensor[0].numpy())
print("Pred  :", y_pred.numpy())
print("Ideal :", Y_tensor[0].numpy())
```

## Analysez model 

```python
y_pred = model(X_tensor)
Y_tensor = Y_tensor
```

```python
import torch
import numpy as np

def kl_divergence(p, q):
    eps = 1e-8
    p = p + eps
    q = q + eps
    return np.sum(p * np.log(p / q))

kl_scores = []

model.eval()
with torch.no_grad():
    preds = model(X_tensor).numpy()
    targets = Y_tensor.numpy()

for i in range(len(preds)):
    kl_scores.append(kl_divergence(targets[i], preds[i]))

print("Mean KL:", np.mean(kl_scores))
## KL = 0  Perfect 
## KL > 0.05  bad 
```

### Total Variation Distance

```python
def tv_distance(p, q):
    return 0.5 * np.sum(np.abs(p - q))

tv_scores = []

for i in range(len(preds)):
    tv_scores.append(tv_distance(targets[i], preds[i]))

print("Mean TV distance:", np.mean(tv_scores))
## between  [0, 1]
# 0   تطابق کامل
# 1   کاملا متفاوت
```

## Fidelity

```python
def fidelity(p, q):
    return (np.sum(np.sqrt(p * q))) ** 2

fid_scores = []

for i in range(len(preds)):
    fid_scores.append(fidelity(targets[i], preds[i]))

print("Mean Fidelity:", np.mean(fid_scores))
# Perfect
```
